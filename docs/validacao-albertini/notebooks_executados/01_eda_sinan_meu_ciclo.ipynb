## 2. Preparação do ambiente e leitura da tabela

Usamos `pandas` para tratamento tabular e `matplotlib` para visualização. O carregamento localiza o CSV tanto quando o kernel é iniciado de `data-science/` quanto de `data-science/notebooks/`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

nome = "sinan_feminino_2015_2024.csv"
candidatos = [
    Path("data-science/dados/brutos") / nome,
    Path("dados/brutos") / nome,
    Path("../dados/brutos") / nome,
]
CAMINHO = next((p for p in candidatos if p.is_file()), None)
if CAMINHO is None:
    raise FileNotFoundError(
        f"CSV não encontrado. Diretório atual: {Path.cwd()}. "
        "Verifique dados/brutos/sinan_feminino_2015_2024.csv"
    )

df = pd.read_csv(CAMINHO, sep=";", encoding="utf-8-sig")
colunas_esperadas = {"ano", "frequencia"}
if not colunas_esperadas.issubset(df.columns):
    raise ValueError(f"Colunas encontradas: {list(df.columns)}; esperadas: {sorted(colunas_esperadas)}")

print("Arquivo:", CAMINHO.resolve())
print("Dimensões:", df.shape)
display(df.head())
```

## 3. Inspeção e qualidade dos dados

Checamos dimensões, tipos, ausências, duplicações e conversão numérica. Como a tabela está agregada por ano, o diagnóstico se refere apenas a **dez linhas anuais**, não à completude dos registros originais do SINAN.

```python
print("Tipos originais:")
display(df.dtypes.rename("tipo").to_frame())
print("Ausências antes da conversão:")
display(df.isna().sum().rename("ausentes").to_frame())
print("Linhas duplicadas:", int(df.duplicated().sum()))

analise = df.copy()
analise["ano"] = pd.to_numeric(analise["ano"], errors="coerce")
analise["frequencia"] = pd.to_numeric(analise["frequencia"], errors="coerce")
print("Ausências após a conversão:")
display(analise.isna().sum().rename("ausentes").to_frame())

if analise[["ano", "frequencia"]].isna().any().any():
    raise ValueError("Há valores inválidos nas colunas numéricas; revisar o CSV.")
if analise["ano"].duplicated().any():
    raise ValueError("Existem anos repetidos; revisar antes de interpretar.")
if (analise["frequencia"] < 0).any():
    raise ValueError("Frequências negativas encontradas.")

analise = analise.sort_values("ano").reset_index(drop=True)
display(analise.describe())
display(analise)
```

```python
# Confere a tabela nacional simplificada com a exportação original preservada.
import io
original = (CAMINHO.parent / "sinannet_cnv_violebr211340177_76_178_23.csv").read_text(encoding="latin-1")
linhas_original = original.splitlines()
inicio = next(i for i, linha in enumerate(linhas_original) if linha.startswith('"Ano da Notific";'))
nacional_original = pd.read_csv(io.StringIO("\n".join(linhas_original[inicio:inicio + 11])), sep=";")
nacional_original.columns = ["ano", "frequencia"]
pd.testing.assert_frame_equal(df.reset_index(drop=True), nacional_original.reset_index(drop=True))
print("Tabela nacional confere com os dez anos da exportação original.")
```

## 4. Exploração descritiva — evolução temporal

Visualizamos a frequência de notificações por ano, sem afirmar que a trajetória representa a incidência real de violência na população.

```python
fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(analise["ano"], analise["frequencia"], marker="o", linewidth=2)
ax.set_title("Notificações de violência — sexo feminino, Brasil (2015–2024)")
ax.set_xlabel("Ano da notificação")
ax.set_ylabel("Frequência de notificações")
ax.set_xticks(analise["ano"].astype(int))
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

primeiro = analise.iloc[0]
ultimo = analise.iloc[-1]
variacao = (ultimo["frequencia"] / primeiro["frequencia"] - 1) * 100
print(f"Primeiro ano: {int(primeiro['ano'])} — {int(primeiro['frequencia']):,} notificações")
print(f"Último ano: {int(ultimo['ano'])} — {int(ultimo['frequencia']):,} notificações")
print(f"Variação do primeiro ao último ano: {variacao:.1f}%")
display(analise.assign(variacao_percentual=analise["frequencia"].pct_change().mul(100).round(2)))
```

```python
tabela_frequencias = analise[["ano", "frequencia"]].copy()

total = tabela_frequencias["frequencia"].sum()

tabela_frequencias["percentual_do_total"] = (
    tabela_frequencias["frequencia"] / total * 100
).round(2)

display(tabela_frequencias)
print(f"Total de notificações no período: {total:,.0f}")
```

```python
minimo = analise["frequencia"].min()
maximo = analise["frequencia"].max()

analise["frequencia_normalizada"] = (
    (analise["frequencia"] - minimo) / (maximo - minimo)
    if maximo != minimo else 0.0
)

display(
    analise[["ano", "frequencia", "frequencia_normalizada"]]
)
```

### Normalização dos dados

Aplicamos a normalização Min-Max às contagens anuais, criando uma
coluna na escala de 0 a 1 e preservando os valores originais.

Essa transformação apenas altera a escala: não representa uma taxa
populacional nem uma estimativa de risco individual. As estatísticas
descritivas e a evolução temporal foram calculadas com as contagens
originais.

```python
frequencias = analise["frequencia"]

resumo = pd.Series({
    "Média anual": frequencias.mean(),
    "Mediana anual": frequencias.median(),
    "Mínimo": frequencias.min(),
    "Máximo": frequencias.max(),
    "Amplitude": frequencias.max() - frequencias.min(),
    "Variância populacional": frequencias.var(ddof=0),
    "Desvio-padrão populacional": frequencias.std(ddof=0),
})

display(resumo.round(2).to_frame("Valor"))

if frequencias.nunique() == len(frequencias):
    print("Não há moda: todas as contagens anuais são diferentes.")
else:
    print("Moda:", frequencias.mode().tolist())
```

## 6. Preparação e análise por UF

Extraímos apenas as 27 linhas de UF (prefixo `..`), excluindo subtotais regionais e Total para evitar dupla contagem. Cada linha possui dez anos e um total, conferido antes da transformação para formato longo. Valores inesperados interrompem a execução: não imputamos ausências nem convertemos símbolos desconhecidos em zero. `uf` contém o nome completo, incluindo Distrito Federal.

```python
from pathlib import Path

nome_arquivo = "sinan_feminino_uf_2015_2024_tabnet.txt"

candidatos = [
    Path("../dados/brutos") / nome_arquivo,
    Path("dados/brutos") / nome_arquivo,
    Path("data-science/dados/brutos") / nome_arquivo,
]

caminho_uf = next(
    (caminho for caminho in candidatos if caminho.is_file()),
    None,
)

if caminho_uf is None:
    raise FileNotFoundError(
        f"Arquivo não encontrado. Pasta atual: {Path.cwd()}"
    )

texto_uf = caminho_uf.read_text(encoding="utf-8-sig")

print("Arquivo encontrado:", caminho_uf.resolve())
print("Quantidade de caracteres:", len(texto_uf))
print("\nInício do conteúdo:\n")
print(texto_uf[:600])
```

```python
import re
import pandas as pd

# Cada linha de UF contém dez valores anuais e um total.
padrao = r'"\.\.\s*([^"]+)"((?:;\s*\d+){11})(?![\d;])'
registros = re.findall(padrao, texto_uf)

if len(registros) != 27:
    raise ValueError(
        f"Esperávamos 27 UFs, mas encontramos {len(registros)}. "
        "Confira se o texto foi copiado integralmente."
    )

linhas = []

for nome_uf, trecho_numerico in registros:
    valores = [
        int(valor.strip())
        for valor in trecho_numerico.split(";")[1:]
    ]

    frequencias = valores[:10]
    total_informado = valores[10]

    if sum(frequencias) != total_informado:
        raise ValueError(f"Total inconsistente para {nome_uf}.")

    for ano, frequencia in zip(range(2015, 2025), frequencias):
        linhas.append({
            "uf": nome_uf.strip(),
            "ano": ano,
            "frequencia": frequencia,
        })

base_uf = pd.DataFrame(linhas).sort_values(
    ["uf", "ano"]
).reset_index(drop=True)

if base_uf.duplicated(["uf", "ano"]).any():
    raise ValueError("Foram encontradas combinações UF/ano repetidas.")

# Compara com a tabela nacional carregada e conferida com o original.
referencia = analise.set_index("ano")["frequencia"].rename("total_nacional")

conferencia = base_uf.groupby("ano")["frequencia"].sum().to_frame(
    "soma_ufs"
)
conferencia["total_nacional"] = referencia
conferencia["confere"] = (
    conferencia["soma_ufs"] == conferencia["total_nacional"]
)

if not conferencia["confere"].all():
    raise ValueError("A soma das UFs difere da referência nacional.")

print("Dimensões:", base_uf.shape)
print("UFs:", base_uf["uf"].nunique())
print("Total de notificações:", base_uf["frequencia"].sum())

display(base_uf.head())
display(conferencia)
assert base_uf.shape == (270, 3)
assert not base_uf.isna().any().any()
assert base_uf["uf"].nunique() == 27
assert base_uf.groupby("uf")["ano"].nunique().eq(10).all()
assert base_uf["frequencia"].ge(0).all()
# Confere também o Total da consulta UF, sem somar subtotais regionais.
total_raw = re.search(r'"Total"((?:;\s*\d+){11})(?![\d;])', texto_uf)
assert total_raw is not None
valores_total = [int(v) for v in total_raw.group(1).split(";")[1:]]
assert valores_total[:10] == referencia.sort_index().tolist()
assert valores_total[10] == int(base_uf["frequencia"].sum())
```

```python
pasta_tratados = caminho_uf.parent.parent / "tratados"
pasta_tratados.mkdir(parents=True, exist_ok=True)

arquivo_tratado = (
    pasta_tratados / "sinan_feminino_uf_2015_2024.csv"
)

base_uf.to_csv(
    arquivo_tratado,
    sep=";",
    index=False,
    encoding="utf-8",
)

print("Base tratada salva em:", arquivo_tratado.resolve())
```

```python
import matplotlib.pyplot as plt

dados_2024 = (
    base_uf.loc[base_uf["ano"] == 2024]
    .sort_values("frequencia")
)

fig, ax = plt.subplots(figsize=(10, 10))

ax.barh(
    dados_2024["uf"],
    dados_2024["frequencia"],
    color="#a52d69",
)

ax.set_title(
    "Notificações de violência interpessoal/autoprovocada\n"
    "Sexo feminino — por UF de notificação, 2024"
)
ax.set_xlabel("Número de notificações")
ax.set_ylabel("UF de notificação")
ax.set_axisbelow(True)
ax.grid(axis="x", alpha=0.25)

fig.text(
    0.01, 0.01,
    "Fonte: SINAN/DATASUS. Contagens absolutas; não representam risco individual.",
    fontsize=9,
)

plt.tight_layout(rect=(0, 0.04, 1, 1))
plt.show()
```
